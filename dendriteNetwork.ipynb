```python
def addDend(G,node,childrenCount,maxChildren=2):
    #think about positions, a vector pointing away from other nodes 
    nodeName = len(G.nodes())
    if childrenCount[node]<maxChildren:
        G.add_edge(node,nodeName)
        childrenCount[node] +=1
        childrenCount[nodeName] = 0

G = nx.Graph()
G.add_node(0)
childrenCount = {0:0}
for _ in range(100):
    target = np.random.randint(len(G.nodes))
    addDend(G,target,childrenCount,maxChildren=10)

nx.draw(G,node_size=100)
```
