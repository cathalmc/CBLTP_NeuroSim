Simulating hodgkin huxley neurons in C++
This is just a quick bit of testing, but may provide a base for an improved version of the simulator

Basic features:
Single compartment conductance based neurons
Voltage and ligand gating 
Simple conductance based synapses

Conductance based neurons:
Resting membrane potential defined in terms of membrane conductances for each ion # Nernst potentials for Na, K, Cl, and Ca, and 
Monoexponential gating defined in terms of Vhalf, Vk, theta0 theta1 sig0 k0



Extra features:
Multicompartments
ChatGPT compile python to c++
Delays
Gap junctions 
Short term plasticity
Long term calcium based plasticity 

```python
import numpy as np
import matplotlib.pyplot as plt

import os,sys
from copy import deepcopy

build_dir,configType = "build", "Debug"
moduleDir = os.path.join(build_dir, configType)
if moduleDir not in sys.path:
    sys.path.append(moduleDir)

import cathalTest

#need to map gates to ion channels to neurons
#although if we assume 2 gates per neuron, we can just map gates to neuron 

#we use inverse slope, 1/k, whereas most of the literature uses k. This is because we want to avoid division in the sigmoid function
def genGate(vHalf,vKi,tau0,tau1,th,tkInv,x0=0.0,exponent=1,minOpenness=0):
    return {"x":x0, "vHalf":vHalf,"vKi":vKi,"tau0":tau0,"tau1":tau1,"th":th,"tkInv":tkInv,"exponent":exponent,"minOpenness":minOpenness}

def genOpenGate():
    return genGate(-1000,1000,100,1000,0,0,x0=1)


```

```python

# gate1 = genGate(-40,2,10,20,-45,10)
# gate2 = genGate(-35,-2,1,10,-45,10)

# ionChannelTypes = {"Ion":0,
#                    "Ca":1,
#                    }
Nav1x6 = {"name":"Nav1x6",
            "ion":"Na",
           "Erev": 90,
           "conductance": 40,
           "CaCurrent": 1, #its not a Ca current, this is just for testing (although, many cation channels will let in a tiny bit of Ca)
           "gates":{
           "vGate1":genGate(-40,-1/4,0.2,2,-20,-1/5,exponent=3,minOpenness=0.0), #inactivation, slow to fast
           "vGate2":genGate(-30,1/4,0.1,0.1,-1000,100), #activation , slow to fast 
           "CaGate":genOpenGate(),
           "spikeGate":genOpenGate()}}


Kv2x1 = {"name":"Kv2x1",
        "ion":"K",
           "Erev": -90,
           "conductance": 20,
           "CaCurrent": 0,
           "gates":{
           "vGate1":genGate(-20,-1/10,1,45,-46,1/44,exponent=4),
           "vGate2":genOpenGate(),
           "CaGate":genOpenGate(),
           "spikeGate":genOpenGate()}}

Kv2x2 = {"name":"Kv2x2",
        "ion":"K",
           "Erev": -90,
           "conductance": 20,
           "CaCurrent": 0,
           "gates":{
           "vGate1":genGate(-30,1/10,1,5,-20,-1/10,exponent=1),
           "vGate2":genGate(-30,-1/10,1,5,-46,1/20,exponent=1),
           "CaGate":genOpenGate(),
           "spikeGate":genOpenGate()}}

Kv3x1 = {"name":"Kv3x1",
        "ion":"K",
           "Erev": -90,
           "conductance": 4,
           "CaCurrent": 0,
           "gates":{
           "vGate1":genGate(-10,-1/8,1,5,-8,1/10,exponent=4,minOpenness=0.0),
           "vGate2":genOpenGate(),
           "CaGate":genOpenGate(),
           "spikeGate":genOpenGate()}}

leakChannel = {"name":"leak",
            "ion":"all",
           "Erev": -75,
           "conductance": 0.5,
           "current":0,
           "CaCurrent": 0,
           "gates":{
           "vGate1":genOpenGate(),
           "vGate2":genOpenGate(),
           "CaGate":genOpenGate(),
           "spikeGate":genOpenGate()}}

def genSynapseGate(xt1,TS):
    xt0 = TS
    xt1 = xt1-xt0
    return genGate(0.5,100,xt0,xt1,0.5,-100)

def genAMPA(TS):
    #for the synapses to work properly you need to know TS
    #This lets the conductance go instantly from 0 to 1
    #I wonder if this will cause instability?
    #If this doesnt work I'll have to implement a more expensive second order ODE
    AMPA = {"name":"AMPAR",
                "ion":"Cation",
            "Erev": 0,
            "conductance": 10,
            "current":0,
            "CaCurrent": 0,
            "gates":{
            "vGate1":genOpenGate(),
            "vGate2":genOpenGate(),
            "CaGate":genOpenGate(),
            "spikeGate":genSynapseGate(3,TS)}}
    return AMPA

def genChannelData(nNeurons,edgeList,TS):
    allChannelData = {key:[] for key in genOpenGate().keys()}
    allChannelData["postNeuron"] = []
    allChannelData["preNeuron"] = []
    allChannelData["conductance"] = []
    #allChannelData["ionIndex"] = [] #could use Erev either?
    allChannelData["Erev"] = []
    allChannelData["nChannels"] = 0
    allChannelData["CaCurrent"] = []

    def addChannelToNeuron(neuronId,ionChannelInfo,allChanData):
        gates = ionChannelInfo["gates"]
        gateSubKeys = gates[list(gates.keys())[0]].keys()
        #pull out the data into a list
        data = {gsk: [gv[gsk] for gv in gates.values()] for gsk in gateSubKeys}
        for key,newData in data.items():
            allChanData[key].append(newData)
        allChannelData["postNeuron"].append(neuronId)
        allChannelData["preNeuron"].append(neuronId)#if its not a synapse, we could make it read its own spike
                                                    #its an open channel anyway so it doesnt matter, and reading its own spike
                                                    #should be the most memory efficient operation
        
        allChannelData["Erev"].append(ionChannelInfo["Erev"])
        allChannelData["conductance"].append(ionChannelInfo["conductance"])
        allChannelData["nChannels"]+=1
        allChannelData["CaCurrent"].append(ionChannelInfo["CaCurrent"])
        
    def addSynapse(preNeuron,postNeuron,synapseInfo,allChanData):
        gates = synapseInfo["gates"]
        gateSubKeys = gates[list(gates.keys())[0]].keys()
        #pull out the data into a list
        data = {gsk: [gv[gsk] for gv in gates.values()] for gsk in gateSubKeys}
        for key,newData in data.items():
            allChanData[key].append(newData)

        allChannelData["postNeuron"].append(postNeuron)
        allChannelData["preNeuron"].append(preNeuron)
        
        allChannelData["Erev"].append(synapseInfo["Erev"])
        allChannelData["conductance"].append(synapseInfo["conductance"])
        allChannelData["CaCurrent"].append(synapseInfo["CaCurrent"])
        allChannelData["nChannels"]+=1

    for i in range(nNeurons):
        addChannelToNeuron(i,leakChannel,allChannelData)
        addChannelToNeuron(i,Nav1x6,allChannelData)
        #addChannelToNeuron(i,Kv2x1,allChannelData)
        addChannelToNeuron(i,Kv2x2,allChannelData)

    for (e0,e1) in edgeList:
        assert e0<nNeurons and e1<nNeurons, f"edge out of range {e0},{e1}"
        addSynapse(e0,e1,genAMPA(TS),allChannelData)

    allChannelData["voltages"] = -70*np.ones(nNeurons,dtype = np.float32) + 5*np.arange(0,nNeurons,dtype=np.float32)
    allChannelData["calciumVals"] = np.zeros(nNeurons,dtype = np.float32)
    allChannelData["spikeVals"] = np.zeros(nNeurons,dtype = np.float32)
    allChannelData["netCurrent"] = np.zeros(nNeurons,dtype = np.float32)
    intKeys = ["preNeuron","postNeuron", "nChannels"]
    for key in allChannelData.keys():
        val = allChannelData[key]
        if key in intKeys: arrDtype = np.int32
        else: arrDtype = np.float32

        if type(val) in [list,tuple]:
            allChannelData[key] = np.array(val,dtype = arrDtype)
        elif type(val)  == np.ndarray:
            pass
        else:
            print(f"{key} is not a list or tuple")
    
    return allChannelData


def sigmoid(z,h,kInv,minOp):
    return minOp +  (1.0-minOp)/(1.0 + np.exp((h-z)*kInv))

def solveGate(x,s,TS,xh,xk,xt0,xt1,xth,xtk,minOpenness):
    xInf = sigmoid(s,xh,xk,minOpenness)#*(xt0+xt1)/TS
    tau = xt0 + xt1*sigmoid(s,xth,xtk,0)
    return x + TS*(xInf-x)/tau
    #return (tau*x + TS*xInf)/(tau+TS)

def solveSystemStandard(nNeurons,ac,TS,runTime,stimInfo={}):
    voltage = ac["voltages"]
    calciumVals = ac["calciumVals"]
    spikeVals = ac["spikeVals"]
    nIterations = int(runTime/TS)
    tv = np.linspace(0,runTime,nIterations)
    vOverTime = [deepcopy(voltage)]
    CaOverTime = [deepcopy(calciumVals)]
    nChannels = ac["nChannels"]
    gateData =[]
    allNetCurrents = []
    for iteration,t in enumerate(tv):   
        #calculate currents
        netCurrent = np.zeros([nNeurons])
        netCaCurrent = np.zeros([nNeurons])
        spikeOverwriteList = np.zeros(nChannels)
        if iteration in stimInfo.keys():
            for syn in stimInfo[iteration]:
                spikeOverwriteList[syn] = 1

        #update the gates
        for i in range(nChannels):
            postSynapticNeuron = ac["postNeuron"][i]
            V = voltage[postSynapticNeuron]
            Ca = calciumVals[postSynapticNeuron]
            preSynapticNeuron = ac["preNeuron"][i]
            spikes = max(spikeVals[preSynapticNeuron],spikeOverwriteList[i])
            xs = ac["x"][i]
            xhs,xks = ac["vHalf"][i],ac["vKi"][i]
            xts0,xts1 = ac["tau0"][i],ac["tau1"][i]
            xths,xtks = ac["th"][i],ac["tkInv"][i]
            exponents = ac["exponent"][i]
            minOpenness = ac["minOpenness"][i]
            Erev = ac["Erev"][i]
            g = ac["conductance"][i]
            CaCurrent = ac["CaCurrent"][i]

            tempTS = TS
            gatingTotal = 1

            for j in range(len(xs)):
                x = xs[j]
                n = exponents[j]
                gatingTotal *= (x**n)

            I = -g*gatingTotal*(V-Erev) 
            netCurrent[postSynapticNeuron] += I
            netCaCurrent[postSynapticNeuron] += CaCurrent*I

            xs[0] = solveGate(xs[0],V,tempTS,xhs[0],xks[0],xts0[0],xts1[0],xths[0],xtks[0],minOpenness[0]) #gate1
            xs[1] = solveGate(xs[1],V,tempTS,xhs[1],xks[1],xts0[1],xts1[1],xths[1],xtks[1],minOpenness[1]) #gate2
            xs[2] = solveGate(xs[2],Ca,tempTS,xhs[2],xks[2],xts0[2],xts1[2],xths[2],xtks[2],minOpenness[2]) #CaGate
            xs[3] = solveGate(xs[3],spikes,tempTS,xhs[3],xks[3],xts0[3],xts1[3],xths[3],xtks[3],minOpenness[3]) #spikeGate


        gateData.append(deepcopy(ac["x"]))
        allNetCurrents.append(deepcopy(netCurrent))

        for i in range(nNeurons):
            voltage[i] += TS*netCurrent[i]
            calciumVals[i] += TS*(0.005*netCaCurrent[i] + (-calciumVals[i]+ 0.001)/2) #This requires some explaination:
                            #We need to convert Ca current (in uA/cm^2) to Ca concentration (in mol/L) so I hardcode the conversion factor (which I just made up TODO:fix)
                            #Cellular Ca is tightly regulated to ~uM concentrations, so I've included a fixed point at 0.001.
                            #I've set the timeconstant for Ca regulation to be 2ms
                            #With this implementation its possible to have 1. calcium efflux (which is fine), and 2. negative Ca concentrations (which is not fine)
                            #but negative Ca would require persistent Vm>140mV which is unphysiological anyway. 
            spikeVals[i] = float(voltage[i] > 0) #we could add delays by making spikes be some dynamical system triggered by voltage > 0. (do a phase portrait to figure it out)
            
        CaOverTime.append(calciumVals.copy())
        vOverTime.append(voltage.copy())
    return np.array(vOverTime),np.array(gateData),np.array(allNetCurrents),np.array(CaOverTime)


def solveSystemCpp(nNeurons,ac,TS,runTime):
    nIterations = int(runTime/TS)
    tv = np.linspace(0,runTime,nIterations)
    cppGateData = []
    allNetCurrents = []
    vs = [deepcopy(ac["voltages"])]
    for i in range(nIterations):
        cathalTest.solveNeuralSystem(ac,2,TS)
        v = deepcopy(ac["voltages"])
        vs.append(v)
        cppGateData.append(deepcopy(ac["x"]))
        allNetCurrents.append(deepcopy(ac["netCurrent"]))

    return np.array(vs),np.array(cppGateData),np.array(allNetCurrents)




runTime = 10
TS = 0.01
nNeurons = 5

edgeList = [[0,1],[1,2],[2,3],[3,4],[4,0]]

allChannelDataPy = genChannelData(nNeurons,edgeList,TS)
stimTime = 5
stimInd = int(stimTime/TS)
stimInfo = {stimInd:[allChannelDataPy["nChannels"]-1]}
#allChannelDataCpp = genChannelData(nNeurons,edgeList,TS)
```

```python
def examineGate(gateInfo,valRange,gateName,ax=None):
    gRange = [min(valRange),max(valRange)]
    vH,vK = gateInfo["vHalf"],gateInfo["vKi"]
    tau0,tau1,th,tkInv = gateInfo["tau0"],gateInfo["tau1"],gateInfo["th"],gateInfo["tkInv"]
    minOp = gateInfo["minOpenness"]
    

    xInf = sigmoid(valRange,vH,vK,minOp)
    tau = tau0 + tau1*sigmoid(valRange,th,tkInv,0)
    if ax == None:
        fig,ax = plt.subplots(2,1,figsize=(8,6), dpi=100)
    else:
        ax[0].plot(valRange,xInf)
        ax[0].axvline(vH,color="r",linestyle="--")
        ax[0].set_title(f"{gateName} xInf")
        ax[0].set_xlim(gRange)
        #x[0].set_ylabel("Gate opening")
        ax[0].set_ylim(-0.01,1.01)

        ax[1].plot(valRange,tau)
        ax[1].axvline(th,color="r",linestyle="--")
        ax[1].set_title(f"{gateName}tau")
        #ax[1].set_ylabel("Time constant (ms)")
        ax[1].set_xlim(gRange)
    
    return xInf

def examineChannel(channel,s0=-100,s1=100):
    fig,ax = plt.subplots(2,4,figsize=(12,8), dpi=80)
    fig.suptitle(f"Channel: {channel['name']}, ion: {channel['ion']}")
    valRange = np.linspace(s0,s1,501)
    netOpening = 1
    for i,(gate,gateData) in enumerate(channel["gates"].items()):
        exponent = gateData["exponent"]
        xInf = examineGate(gateData,valRange,gate,[ax[0,i],ax[1,i]])
        netOpening *= xInf**exponent

    fig,ax = plt.subplots(1,1,dpi=90,figsize=[7,4])
    ax.plot(valRange,netOpening)



amp = genAMPA(TS)

examineChannel(amp,0,1)
```

```python
#need to be able to test this in a reduced simulation 
def temporalChannelTest(channel,duration,TS):
    channel = deepcopy(channel)
    nIterations = int(duration/TS)
    voltages = -80*np.ones(nIterations)
    inputs = np.zeros(nIterations)
    ind = int(nIterations/2)
    inputs[ind:(ind+1)] = 1
    ind = int(3*nIterations/4)
    inputs[ind:(ind+1)] = 1
    currents = np.zeros(nIterations)
    gatingTotals = np.zeros(nIterations)
    Erev = channel["Erev"]
    g = channel["conductance"]
    tv=np.linspace(0,duration,nIterations)
    for it in range(nIterations):
        gatingTotal = 1
        for gate,val in channel["gates"].items():
            gatingTotal *= (val["x"]**val["exponent"])

        V = voltages[it]
        I = -g*gatingTotal*(V-Erev) 
        currents[it]=I
        gatingTotals[it] = gatingTotal
        #update gates,
        #then update voltages
        gateDataMap = {"vGate1":V,"vGate2":V,"CaGate":0,"spikeGate":inputs[it]}
        for gate,val in channel["gates"].items():
            d = gateDataMap[gate]
            val["x"] = solveGate(val["x"],d,TS,val["vHalf"],val["vKi"],val["tau0"],val["tau1"],val["th"],val["tkInv"],val["minOpenness"])

    return currents,gatingTotals,tv


currents,gatingTotals,tv = temporalChannelTest(amp,50,0.01,)
fig,ax = plt.subplots(2,1,dpi=90,sharex=True)
ax[0].plot(tv,currents)
ax[0].set_title("current")
ax[1].plot(tv,gatingTotals)
ax[1].set_title("gating value")
```

```python

def examineGate(gateInfo,gRange,gateName,ax=None):
    sps = np.linspace(gRange[0],gRange[1],501)
    vH,vK = gateInfo["vHalf"],gateInfo["vKi"]
    tau0,tau1,th,tkInv = gateInfo["tau0"],gateInfo["tau1"],gateInfo["th"],gateInfo["tkInv"]

    xInf = sigmoid(sps,vH,vK)
    tau = tau0 + tau1*sigmoid(sps,th,tkInv)
    if ax == None:
        fig,ax = plt.subplots(2,1,figsize=(8,6), dpi=100)
    else:
        ax[0].plot(sps,xInf)
        ax[0].axvline(vH,color="r",linestyle="--")
        ax[0].set_title(f"{gateName} xInf")
        ax[0].set_xlim(gRange)
        #x[0].set_ylabel("Gate opening")
        ax[0].set_ylim(-0.01,1.01)

        ax[1].plot(sps,tau)
        ax[1].axvline(th,color="r",linestyle="--")
        ax[1].set_title(f"{gateName}tau")
        #ax[1].set_ylabel("Time constant (ms)")
        ax[1].set_xlim(gRange)
        

def examineChannel(channel):
    fig,ax = plt.subplots(2,4,figsize=(12,8), dpi=100)
    fig.suptitle(f"Channel: {channel['name']}, ion: {channel['ion']}")
    for i,(gate,gateData) in enumerate(channel["gates"].items()):
        examineGate(gateData,[-100,100],gate,[ax[0,i],ax[1,i]])

examineChannel(leakChannel)
examineChannel(Nav1x6)
examineChannel(Kv2x1)
examineChannel(Kv3x1)
```

```python

def solveGate(x,s,TS,xh,xk,xt0,xt1,xth,xtk):
    xInf = sigmoid(s,xh,xk)#*(xt0+xt1)/TS
    tau = xt0 + xt1*sigmoid(s,xth,xtk)
    return (tau*x + TS*xInf)/(tau+TS)





def runGateSim(xt0,xt1,TS):
    #TS = 0.02
    x = 0
    xh = 0.5
    xk = 100.0
    xt0 = TS/1000
    xt1 = xt1-xt0
    xth = 0.5
    xtk = -100.0
    runTime = 40
    
    nIterations = int(runTime/TS)
    tv = np.linspace(0,runTime,nIterations)
    xOverTime = []
    for i in range(nIterations):
        s = 0
        if i == int(5/TS):
            s = 1
        x = solveGate(x,s,TS,xh,xk,xt0,xt1,xth,xtk)
        xOverTime.append(x)

    fig,ax = plt.subplots(1,1,figsize=(8,6), dpi=100, sharex=True)
    ax.plot(tv,xOverTime)

from ipywidgets import interact
interact(runGateSim,xt0=(0.01,1.0,0.01),xt1=(0.1,10.0,0.1),TS=(0.01,0.1,0.01))
```
